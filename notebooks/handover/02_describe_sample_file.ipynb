# Describe One Sample Artifact

This notebook inspects one chosen artifact for one sample date. It is meant to stay lean: metadata, preview rows or features, one spatial slice, and one histogram.

```python
from pathlib import Path
import sys

candidate_dirs = [
    Path.cwd(),
    Path.cwd() / 'notebooks' / 'handover',
]
for candidate in candidate_dirs:
    if (candidate / 'sample_data_helpers.py').exists():
        sys.path.insert(0, str(candidate))
        break

from sample_data_helpers import build_artifact_paths, inspect_artifact, render_artifact_view
```

```python
# Parameters
SAMPLE_DATE = '20230101'
TARGET_ENV = 'neptune'  # 'azure' or 'neptune'
TARGET_ARTIFACT = 'pt_preprocessed'

ROOT_OVERRIDES = {}
```

```python
paths = build_artifact_paths(SAMPLE_DATE, TARGET_ENV, ROOT_OVERRIDES)
artifact_path = paths[TARGET_ARTIFACT]
artifact_path
```

```python
if artifact_path is None or not artifact_path.exists():
    raise FileNotFoundError(f'Missing artifact: {TARGET_ARTIFACT} for {SAMPLE_DATE} in {TARGET_ENV}: {artifact_path}')

view = inspect_artifact(artifact_path, TARGET_ARTIFACT)
render_artifact_view(view)
```
