# Compare The Same Sample Across Representations

This notebook compares the same `WAVEANYYYYMMDD` sample across raw `.nc`, engineered `.parquet`, and training-ready `.pt` when those files exist.

```python
from pathlib import Path
import sys

candidate_dirs = [
    Path.cwd(),
    Path.cwd() / 'notebooks' / 'handover',
]
for candidate in candidate_dirs:
    if (candidate / 'sample_data_helpers.py').exists():
        sys.path.insert(0, str(candidate))
        break

from sample_data_helpers import build_artifact_paths, compare_artifacts, inspect_artifact, render_artifact_view
```

```python
# Parameters
SAMPLE_DATE = '20230101'
TARGET_ENV = 'azure'  # 'azure' or 'neptune'
ARTIFACT_KEYS = [
    'nc_raw_uncorrected',
    'nc_reference',
    'parquet_engineered',
    'pt_preprocessed_blob',
]

ROOT_OVERRIDES = {}
```

```python
comparison_df = compare_artifacts(
    sample_date=SAMPLE_DATE,
    env=TARGET_ENV,
    artifact_keys=ARTIFACT_KEYS,
    root_overrides=ROOT_OVERRIDES,
)
comparison_df
```

```python
paths = build_artifact_paths(SAMPLE_DATE, TARGET_ENV, ROOT_OVERRIDES)
for artifact_key in ARTIFACT_KEYS:
    artifact_path = paths.get(artifact_key)
    if artifact_path is None or not artifact_path.exists():
        print(f'Skipping missing artifact: {artifact_key} -> {artifact_path}')
        continue

    print('\n' + '=' * 80)
    view = inspect_artifact(artifact_path, artifact_key)
    render_artifact_view(view)
```
